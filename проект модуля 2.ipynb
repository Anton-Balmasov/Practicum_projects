### 2.1. Группы по срокам финансирования

Разделите все компании на три группы:

* Единичное финансирование — был всего один раунд финансирования.

* Срок финансирования до года — между первым и последним раундом финансирования прошло не более года.

* Срок финансирования более года.

Визуализируйте соотношение этих групп, создав два графика:

* По количеству компаний: Покажите, какой процент от общего числа компаний относится к каждой из трёх групп.
* По объёму инвестиций: Отобразите, какую долю от общего объёма привлечённых средств получила каждая группа.

Совет: Для ясности и согласованности используйте единую цветовую палитру для всех графиков, чтобы каждая категория (например, «Единичное финансирование») всегда отображалась одним цветом.

```python
def categorize_company(row):
    if row['funding_rounds'] == 1:
        return 'Единичное финансирование'
    elif (row['last_funding_at'] - row['first_funding_at']).days <= 365:
        return 'Срок финансирования до года'
    else:
        return 'Срок финансирования более года'

cbi['category'] = cbi.apply(categorize_company, axis=1)
cbi['category']
```

```python
# Агрегируем данные по категориям
grouped_data = cbi.groupby('category').size().reset_index(name='company_count')

# Общее количество компаний
total_companies = cbi.shape[0]

# Процент компаний в каждой группе
grouped_data['percentage'] = grouped_data['company_count'] / total_companies * 100

# Визуализация
plt.figure(figsize=(10, 6))
sns.barplot(x='category', y='percentage', data=grouped_data,palette=["b", "g", "y"])
plt.title('Соотношение компаний по категориям финансирования')
plt.xlabel('Категория')
plt.ylabel('Процент компаний (%)')
plt.show()
print(total_companies)
print(grouped_data)
```

```python


# Группировка данных по категориям и суммирование инвестиций
grouped_data =cbi.groupby('category').agg(total_funding=('funding_total_usd', 'sum'))

# Общие инвестиции
total_funding = grouped_data['total_funding'].sum()

# Нормируем данные (рассчитываем долю каждой группы)
grouped_data['funding_percentage'] = grouped_data['total_funding'] / total_funding * 100
grouped_data_sorted = grouped_data.sort_values(by='funding_percentage', ascending=False)

print(grouped_data_sorted )

# Визуализация
plt.figure(figsize=(10, 6))
sns.barplot(x=grouped_data_sorted.index, y=grouped_data_sorted['funding_percentage'],palette=[ "g","b", "y"])
plt.title('Доли привлеченных инвестиций по категориям финансирования')
plt.xlabel('Категория')
plt.ylabel('Доля привлеченных инвестиций (%)')
plt.show()
```

На графиках мы видим что единичное финансирвание получет большинство компаний - 58.9%
А срок финансирования более года привлек -61.8% инвестиций.

Определите компании с аномальным объёмом общего финансирования — используйте метод IQR отдельно по каждому сегменту. Напомним, что все нишевые сегменты должны быть объединены в одну группу, а средние — в другую.

Определите сегменты рынка с наибольшей долей компаний, получивших аномальное финансирование, и выведите топ таких сегментов.

```python

# Если ещё не сделано — приводим funding к числу
cbi['funding_total_usd'] = pd.to_numeric(cbi['funding_total_usd'], errors='coerce')
# Заменяем NaN в market на "Unknown" для группировки
cbi['market_filled'] = cbi['market'].fillna('Unknown')

def flag_outliers_iqr(group):
    Q1 = group.quantile(0.25)
    Q3 = group.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return (group < lower_bound) | (group > upper_bound)

# Применяем к каждой группе по market_filled
cbi['is_outlier'] = cbi.groupby('market_filled')['funding_total_usd'].apply(flag_outliers_iqr)

print(f"Компании с аномальным финансированием: ")


#Выведем компании с аномальным финансированием:
anomalies = cbi[cbi['is_outlier']].sort_values(['market_filled', 'funding_total_usd'], ascending=False)
display(anomalies[['name', 'funding_total_usd', 'market_filled']])

outlier_ratio = cbi.groupby('market_filled').agg(
    total_companies=('is_outlier', 'count'),
    outlier_count=('is_outlier', 'sum')
).reset_index()

print(f" Cегменты рынка с наибольшей долей компаний, получивших аномальное финансирование, топ таких сегментов:")

# Доля аномалий в %
outlier_ratio['outlier_ratio_%'] = (outlier_ratio['outlier_count'] / outlier_ratio['total_companies']) * 100
outlier_ratio['outlier_ratio_%'] = outlier_ratio['outlier_ratio_%'].round(2)

top_outlier_segments = outlier_ratio.sort_values('outlier_ratio_%', ascending=False).head(10)

display(top_outlier_segments)
```

### 4.3 Годовая динамика доли возвращённых средств по типам финансирования

Заказчик хочет знать, какая часть вложенных или выданных денег со временем возвращается обратно инвесторам или финансистам. Ваша цель — для каждого года и каждого вида финансирования рассчитать нормированные значения возврата средств: то есть какую долю возвращённые средства составляют от предоставленных. При этом слишком большие аномальные значения, то есть неадекватные выбросы, нужно заменить на пропуски.

Совет: когда будете делить сумму возвращённых средств на суммарный объём привлечённого финансирования по конкретному году, добавьте к знаменателю небольшое число, например `1e-60`. Это поможет избежать деления на ноль.

```python
data=(cbi_filtered[['funding_year','seed',
'venture',
'equity_crowdfunding',
'undisclosed',
'convertible_note',
'debt_financing',
'angel',
'grant',
'private_equity',
'post_ipo_equity',
'post_ipo_debt',
'secondary_market',
'product_crowdfunding' ]])
```

```python

# Группируем по funding_year и суммируем все числовые столбцы
result2 = data.groupby('funding_year').sum()/1000000
result2 = result2.round(2)
# Выводим результат
display(result2)
```

```python
# выполняем деление
result = cbr.div(result2+(1e-60 ))
result = result.round(2)
display(result)
```

```python
# Задаём порог для аномальных значений
threshold = 100

# Функция для замены аномальных значений на None
def replace_outliers(value):
    if isinstance(value, (int, float)) and not pd.isna(value):
        if abs(value) > threshold:
            return None
        else:
            return value
    

# Применяем функцию ко всем столбцам DataFrame
for column in result.columns:
    result[column] = result[column].apply(replace_outliers)

print("DataFrame после замены аномальных значений:")
display(result)
```
